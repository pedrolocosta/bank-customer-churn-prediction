## Adding custom functions

```python
def std_col_names(original_df:pd.DataFrame):

    '''Apply a standard cleaning procedure to the original DataFrame's column 
    names by converting them to lowercase, removing extra whitespace, and 
    replacing spaces with underscores to follow the `snake_case` naming 
    convention. Then, replace the original column names with the cleaned ones.'''

    old_col_names = original_df.columns.tolist()
    
    new_col_names = []
    for old_name in old_col_names:
        name_stripped = old_name.strip()
        name_lowered = name_stripped.lower()
        name_no_space = name_lowered.replace(' ', '_')
        new_col_names.append(name_no_space)

    original_df.columns = new_col_names

    return print('Lista dos novos nomes das colunas:\n', original_df.columns)
```

```python
def assess_df(dataframe:pd.DataFrame, num_rows:int):

    '''Creates a DataFrame with the number of rows determinated by the 
    'num_row' variable, and sets the last row show the data type of the
     values in that column.'''

    assess_df = pd.concat([
        dataframe.sample(num_rows),
        dataframe.dtypes.to_frame().T.rename(index={0: 'Data Type'})
    ])

    return display(assess_df)
```

## Criar interação entre atributos

```python
numerical_features = [
    'credit_score',
    'age',
    'tenure',
    'balance',
    'num_of_products',
    'has_cr_card',
    'is_active_member',
    'estimated_salary'
]
```

### Planejamento da geração de interação entre atributos

Será implementado o algorítimo Polynomial Features nos três conjuntos de dados. O ajuste será realizado apenas no conjunto de treinamento e a transformação após o ajuste em todos os conjuntos.

### Implementação do plano de ação

```python
poly = PolynomialFeatures(degree=2, include_bias=False)

# Learn category mapping from training data only.
poly.fit(X_train[numerical_features])
polynomial_columns = poly.get_feature_names_out(numerical_features)

X_poly = {}
for name, split_X in [
    ('training', X_train),
    ('validation', X_valid),
    ('test', X_test)
]:
    transformed_data = poly.transform(split_X[numerical_features])

    # Restoring column names and preserving the original row index.
    poly_df = pd.DataFrame(
        transformed_data,
        columns=polynomial_columns,
        index=split_X.index
    )
    # Keeping columns not included in the polynomial expansion.
    other_features = split_X.drop(columns=numerical_features)

    # Storing the complet features set.
    X_poly[name] = pd.concat([other_features, poly_df], axis=1)
```

```python
# Assign encoded feature splits to separate variables
X_train_poly = X_poly['training']
X_valid_poly = X_poly['validation']
X_test_poly = X_poly['test']
```

### Planejamento do processamento das variáveis categóricas

Apenas as variáveis das colunas `geography` e `gender` são não numéricas da categoria nominal. Não há variáveis da categoria ordinal.

```python
# Declaring the categorical features
categorical_features = ['geography', 'gender']
```

Será criado dois novos DataFrames com tratamento diferentes para aplicação em modelos de ML diferentes: 
- No df que será direcionado para modelos não baseados em árvores será aplicado OHE nas colunas `geography` e `gender`.
- No df que será direcionado para modelos baseados em árvores, será aplicado o Label Encoding nas colunas `geography` e `gender`. 

### Implementação do plano de ação

#### Criar df para algorítimos não baseados em árvore

```python
# codificando as colunas categóricas com one-hot encoding

# dicionário para armazenar os conjuntos de features codificados
ohe_X = {}

for name, split_X in [
    ('training', X_train_poly),
    ('validação', X_valid_poly),
    ('teste', X_test_poly)   
]:
    prepared_X = split_X.copy()
    
    # One-hot encode the gender column
    ohe = pd.get_dummies(
        prepared_X, columns=categorical_features, 
        drop_first=True, dtype=int
    )
   
    # Align validation and test columns with training columns
    if name != 'training':
        ohe = ohe.reindex(
            columns=ohe_X['training'].columns,
            fill_value=0
        )

    # Store the encoded feature split in the dictionary
    ohe_X[name] = ohe
```

```python
# Assign encoded feature splits to separate variables
X_train_ohe = ohe_X['training']
X_valid_ohe = ohe_X['validação']
X_test_ohe = ohe_X['teste']
```

#### Checando a implementação realizada

```python
assert X_train_ohe.index.equals(X_train.index)
assert X_valid_ohe.index.equals(X_valid.index)
assert X_test_ohe.index.equals(X_test.index)

for name, split_X in [
    ('Training', X_train_poly),
    ('Treinamento OHE', X_train_ohe),
    ('Validation', X_valid_poly),
    ('Validation OHE', X_valid_ohe),
    ('Test', X_test_poly),
    ('Test OHE', X_test_ohe)
]:
    print(f'{name}: {split_X.shape}')
    print(', '.join(split_X.columns), '\n')
```

```python
assess_df(X_train_ohe, 4)
```

#### Criar df para algorítimo baseado em árvore

```python
label = OrdinalEncoder()

# Learn category mappings from training data only
label.fit(X_train_poly[categorical_features])

label_X = {}
for name, split_X in [
    ('training', X_train_poly),
    ('validation', X_valid_poly),
    ('test', X_test_poly)
]:
    prepared_X = split_X.copy()

    # Replace whole columns with the numeric array.
    prepared_X[categorical_features] = label.transform(
        prepared_X[categorical_features]
    )
    # Store the complet feature set.
    label_X[name] = prepared_X
```

```python
# Assign encoded feature splits to separate variables
X_train_label = label_X['training']
X_valid_label = label_X['validation']
X_test_label = label_X['test']
```

#### Checando a implementação realizada

```python
assert X_train_label.index.equals(X_train.index)
assert X_valid_label.index.equals(X_valid.index)
assert X_test_label.index.equals(X_test.index)

for name, split_X in [
    ('Training', X_train_poly),
    ('Training Labeled', X_train_label),
    ('Validation', X_valid_poly),
    ('Validation Labeled', X_valid_label),
    ('Test', X_test_poly),
    ('Test Labeled', X_test_label)
]:
    print(f'{name}: {split_X.shape}')
    print(', '.join(split_X.columns), '\n')
```

# Improving Model Performance

### Comparing Three Class-Balancing Strategies Against the Baseline

```python
results_balancing = []
trained_model = {}

for model_name, training_features,validation_features in [
    ('Logistic Regression', X_train_scaled, X_valid_scaled),
    ('Random Forest', X_train_label, X_valid_label)
]:
    for approach in [
        'Baseline',
        'Oversampling',
        'Undersampling',
        'Class weighting'
    ]:
        # Start each experiment with the original training set
        features_fit = training_features
        target_fit = y_train
        class_weight = None

        if approach == 'Oversampling':
            features_zeros = training_features[y_train == 0]
            features_ones = training_features[y_train == 1]
            target_zeros = y_train[y_train == 0]
            target_ones = y_train[y_train ==1]

            features_fit = pd.concat([features_zeros] + [features_ones] * round(len(features_zeros) / len(features_ones)))
            target_fit = pd.concat([target_zeros] + [target_ones] * round(len(target_zeros) / len(target_ones)))
            
            features_fit, target_fit = shuffle(features_fit, target_fit, random_state=67)
            
        elif approach == 'Undersampling':
            features_zeros = training_features[y_train == 0]
            features_ones = training_features[y_train == 1]
            target_zeros = y_train[y_train == 0]
            target_ones = y_train[y_train == 1]

            features_fit = pd.concat(
                [features_zeros.sample(
                    frac=(len(features_ones) / len(features_zeros)), random_state=67
                )] + [features_ones]
            )
            
            target_fit = pd.concat(
                [target_zeros.sample(
                    frac=(len(target_ones) / len(target_zeros)), random_state=67
                )] + [target_ones]
            )
                        
            features_fit, target_fit = shuffle(
                features_fit, target_fit, random_state=67
            )
                      
        elif approach == 'Class weighting':
            class_weight = 'balanced'

        # Initialize a fresh model for this experiment
        if model_name == 'Logistic Regression':
            model = LogisticRegression(
                solver='liblinear',
                max_iter=1000,
                random_state=67,
                class_weight=class_weight
            )
        else:
            model = RandomForestClassifier(
                n_estimators=100,
                random_state=67,
                class_weight=class_weight
            )

        model.fit(features_fit, target_fit)

        # Evaluate the trained model on validation data
        predictions = model.predict(validation_features)
        probabilities = model.predict_proba(validation_features)[:, 1]

        results_balancing.append({
            'model': model_name,
            'approach': approach,
            'training rows': len(target_fit),
            'accuracy': accuracy_score(y_valid, predictions),
            'precision': precision_score(y_valid, predictions, zero_division=0),
            'recall': recall_score(y_valid, predictions, zero_division=0),
            'f1': f1_score(y_valid, predictions, zero_division=0),
            'roc_auc': roc_auc_score(y_valid, probabilities)
        })
        
        trained_model[f'{model_name} - {approach}'] = model
```

## Comparing Class-Balancing Results

```python
balancing_results_df = pd.DataFrame(results_balancing)

display(
    balancing_results_df
    .sort_values('f1', ascending=False)
    .round(3)
)
```

### Finding the Best Hyperparameter Configuration

```python
# Prepare hyperparameter candidates
forest_params = []

for est in range(50, 101, 10):
    for depth in [5, 10, None]:
        for leaf in [1,3]:
            forest_params.append({
                'n_estimators': est,
                'max_depth': depth,
                'min_samples_leaf': leaf
            })

logistic_params = [
    {'C': 0.01, 'solver':'liblinear'},
    {'C': 0.1, 'solver':'liblinear'},
    {'C': 1, 'solver':'liblinear'},
    {'C': 10, 'solver':'liblinear'}
]
```

```python
# Initialize tracking variables
tuning_results = []

best_f1 = {
    'Logistic Regression': -1,
    'Random Forest': -1
}

best_models = {}
best_settings = {}
```

```python
# Extend the experiment with hyperparameter search
for model_name, training_features,validation_features in [
    ('Logistic Regression', X_train_scaled, X_valid_scaled),
    ('Random Forest', X_train_label, X_valid_label)
]:
    # Select candidate parameters for each model
    if model_name == 'Logistic Regression':
         params_options = logistic_params
    else:
         params_options = forest_params
    
    for approach in [
        'Baseline',
        'Oversampling',
        'Undersampling',
        'Class weighting'
    ]:
        # Start each experiment with the original training set
        features_fit = training_features
        target_fit = y_train
        class_weight = None

        if approach == 'Oversampling':
            features_zeros = training_features[y_train == 0]
            features_ones = training_features[y_train == 1]
            target_zeros = y_train[y_train == 0]
            target_ones = y_train[y_train ==1]
    
            features_fit = pd.concat([features_zeros] + [features_ones] * round(len(features_zeros) / len(features_ones)))
            target_fit = pd.concat([target_zeros] + [target_ones] * round(len(target_zeros) / len(target_ones)))
    
            features_fit, target_fit = shuffle(features_fit, target_fit, random_state=67)
    
        elif approach == 'Undersampling':
            features_zeros = training_features[y_train == 0]
            features_ones = training_features[y_train == 1]
            target_zeros = y_train[y_train == 0]
            target_ones = y_train[y_train == 1]
    
            features_fit = pd.concat(
                [features_zeros.sample(
                    frac=(len(features_ones) / len(features_zeros)), random_state=67
                )] + [features_ones]
            )
            target_fit = pd.concat(
                [target_zeros.sample(
                    frac=(len(target_ones) / len(target_zeros)), random_state=67
                )] + [target_ones]
            )
                        
            features_fit, target_fit = shuffle(
                features_fit, target_fit, random_state=67
            )
        
        elif approach == 'Class weighting':
            class_weight = 'balanced'

        # Initialize a fresh model for this experiment
        for params in params_options:
             if model_name == 'Logistic Regression':
                  model = LogisticRegression(
                       max_iter=1000,
                       class_weight=class_weight,
                       random_state=67,
                       **params
                  )
             else:
                model = RandomForestClassifier(
                     random_state=67,
                     class_weight=class_weight,
                     **params
                )

             model.fit(features_fit, target_fit)

             # Evaluate the trained model on validation data
             predictions = model.predict(validation_features)
             probabilities = model.predict_proba(validation_features)[:, 1]

             score = f1_score(y_valid, predictions, zero_division=0)

             result = {
                'model': model_name,
                'approach': approach,
                'parameters': params.copy(),
                'f1': score,
                'precision': precision_score(y_valid, predictions, zero_division=0),
                'recall': recall_score(y_valid, predictions, zero_division=0),
                'roc_auc': roc_auc_score(y_valid, probabilities)
             }

             tuning_results.append(result)
                        
             if score > best_f1[model_name]:
                best_f1[model_name] = score
                best_models[model_name] = model
                best_settings[model_name] = result.copy()
```

```python
# Inspect tuning results
tuning_results_df = pd.DataFrame(tuning_results)

display(
    tuning_results_df
    .sort_values('f1', ascending=False, kind='stable')
    .head(10)
    .round(3)
)
```

For Random Forest, balanced class weights improved validation F1-score from **0.581** to **0.634**. The highest validation F1-score was achieved by **Random Forest with balanced class weights**, using `n_estimators=100`, `max_depth=None`, and `min_samples_leaf=1`.

**Metric interpretation:** F1-score is the harmonic mean of precision and recall. Configurations with similar F1-scores can still have different precision–recall trade-offs, which matter for retention decisions.

**ROC-AUC:** The selected Random Forest achieved a validation ROC-AUC of **0.847**, substantially above the random-classifier reference of 0.5. ROC-AUC complements F1-score by measuring the model's ability to rank customers by churn risk across classification thresholds.
